```python
# =============================================================================
# PART 1: BIOLOGICAL vs ARTIFICIAL NEURON
# =============================================================================
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
import warnings
warnings.filterwarnings('ignore')

import plotly.io as pio
try:
    import google.colab  # noqa: F401
    pio.renderers.default = "colab"
except ImportError:
    pio.renderers.default = "notebook"
display(HTML("""
<div class="concept-box">
    <h2 style="margin-top:0; color:#667eea;">ðŸ”¬ The Perceptron: Loosely Inspired by Biology</h2>
    <p>In 1958, Frank Rosenblatt created the <strong>perceptron</strong> â€” taking inspiration from how biological neurons work.</p>
    <p>Let's compare the two side by side!</p>
</div>
"""))

def create_biological_neuron_detailed():
    """Create a detailed, anatomically-inspired biological neuron visualization"""
    
    fig = go.Figure()
    
    np.random.seed(42)
    dendrite_color = '#8B4513'
    
    def draw_branch(start_x, start_y, angle, length, depth, traces):
        if depth == 0 or length < 0.1:
            return
        
        end_x = start_x + length * np.cos(angle)
        end_y = start_y + length * np.sin(angle)
        
        traces.append(go.Scatter(
            x=[start_x, end_x], y=[start_y, end_y],
            mode='lines',
            line=dict(color=dendrite_color, width=max(1, 4 - depth)),
            hoverinfo='skip', showlegend=False
        ))
        
        if depth >= 2:
            traces.append(go.Scatter(
                x=[end_x], y=[end_y],
                mode='markers',
                marker=dict(size=6, color='#D2691E'),
                hoverinfo='text', hovertext='Synapse',
                showlegend=False
            ))
        
        if depth > 0:
            for _ in range(np.random.randint(2, 4)):
                new_angle = angle + np.random.uniform(-0.7, 0.7)
                new_length = length * np.random.uniform(0.5, 0.8)
                draw_branch(end_x, end_y, new_angle, new_length, depth - 1, traces)
    
    dendrite_traces = []
    soma_x, soma_y = 2.5, 2.5
    
    for angle in [2.5, 2.8, 3.1, 3.5, 3.9, 2.2]:
        draw_branch(soma_x, soma_y, angle, 1.2, 3, dendrite_traces)
    
    for trace in dendrite_traces:
        fig.add_trace(trace)
    
    # Soma
    theta = np.linspace(0, 2*np.pi, 100)
    r = 0.5 + 0.08 * np.sin(5*theta) + 0.05 * np.sin(8*theta)
    fig.add_trace(go.Scatter(
        x=soma_x + r * np.cos(theta), y=soma_y + r * np.sin(theta),
        fill='toself', fillcolor='rgba(255, 182, 193, 0.9)',
        line=dict(color='#FF69B4', width=3),
        hoverinfo='text', hovertext='Cell Body (Soma)', showlegend=False
    ))
    
    # Nucleus
    fig.add_trace(go.Scatter(
        x=soma_x + 0.25 * np.cos(theta), y=soma_y + 0.25 * np.sin(theta),
        fill='toself', fillcolor='rgba(138, 43, 226, 0.7)',
        line=dict(color='#9400D3', width=2),
        hoverinfo='text', hovertext='Nucleus', showlegend=False
    ))
    
    # Axon
    axon_x = np.linspace(soma_x + 0.8, soma_x + 4, 50)
    axon_y = soma_y - 0.5 + 0.15 * np.sin(3 * (axon_x - soma_x))
    fig.add_trace(go.Scatter(x=axon_x, y=axon_y, mode='lines',
                             line=dict(color='#4169E1', width=6), showlegend=False))
    
    # Myelin
    for mx in np.linspace(soma_x + 1.2, soma_x + 3.5, 5):
        my = soma_y - 0.5 + 0.15 * np.sin(3 * (mx - soma_x))
        fig.add_trace(go.Scatter(
            x=mx + 0.2 * np.cos(theta), y=my + 0.12 * np.sin(theta),
            fill='toself', fillcolor='rgba(255, 215, 0, 0.6)',
            line=dict(color='#FFD700', width=1), showlegend=False, hoverinfo='skip'
        ))
    
    # Terminals
    term_x = soma_x + 4
    term_y = soma_y - 0.5 + 0.15 * np.sin(3 * (term_x - soma_x))
    for tx, ty in [(term_x + 0.6, term_y + 0.3), (term_x + 0.7, term_y), (term_x + 0.6, term_y - 0.3)]:
        fig.add_trace(go.Scatter(x=[term_x, tx], y=[term_y, ty], mode='lines',
                                 line=dict(color='#4169E1', width=3), showlegend=False))
        fig.add_trace(go.Scatter(x=[tx], y=[ty], mode='markers',
                                 marker=dict(size=15, color='#FF4500'), showlegend=False))
    
    # Labels
    for lx, ly, text, color in [(1.0, 4.2, 'Dendrites', '#8B4513'), (soma_x, soma_y+0.9, 'Soma', '#FF69B4'),
                                 (soma_x+2.5, soma_y+0.3, 'Axon', '#4169E1'), (soma_x+4.8, term_y, 'Terminals', '#FF4500')]:
        fig.add_annotation(x=lx, y=ly, text=f'<b>{text}</b>', showarrow=False,
                          font=dict(size=11, color=color), bgcolor='rgba(255,255,255,0.8)')
    
    fig.update_layout(height=450, width=700, showlegend=False,
                      xaxis=dict(visible=False, range=[0, 8]),
                      yaxis=dict(visible=False, range=[0, 5], scaleanchor='x'),
                      margin=dict(l=20, r=20, t=40, b=20))
    return fig

fig_bio = create_biological_neuron_detailed()
fig_bio.show()

display(HTML("""
<div class="insight-box">
    <h3 style="margin-top:0; color:#856404;">ðŸ’¡ How a Biological Neuron Works</h3>
    <ol style="margin-bottom: 0;">
        <li><strong>Dendrites</strong> receive <em>electrochemical signals</em> from thousands of other neurons</li>
        <li><strong>Soma</strong> integrates all incoming signals</li>
        <li>If the total exceeds a <strong>threshold</strong>, the neuron "fires"</li>
        <li>Signal travels down the <strong>axon</strong> to the terminals</li>
    </ol>
</div>
"""))
```

```python
# =============================================================================
# PART 2: THE ARTIFICIAL NEURON - STEP BY STEP
# =============================================================================

display(HTML("""
<div class="concept-box">
    <h2 style="margin-top:0; color:#667eea;">ðŸ¤– The Artificial Neuron (Perceptron)</h2>
    <p>Now let's see the artificial version - broken into <strong>clear, separate steps</strong>!</p>
</div>
"""))

class PerceptronViz:
    def __init__(self):
        self.create_interface()
    
    def sigmoid(self, x):
        return 1 / (1 + np.exp(-np.clip(x, -500, 500)))
    
    def create_interface(self):
        display(HTML("""
        <div style="background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); 
                    color: white; padding: 20px; border-radius: 12px; text-align: center;
                    margin-bottom: 20px;">
            <h3 style="margin: 0;">ðŸ”§ Perceptron Step-by-Step</h3>
        </div>
        """))
        
        layout = widgets.Layout(width='180px')
        style = {'description_width': '40px'}
        
        self.x1 = widgets.FloatSlider(value=0.6, min=0, max=1, step=0.1, description='xâ‚:', style=style, layout=layout)
        self.x2 = widgets.FloatSlider(value=0.4, min=0, max=1, step=0.1, description='xâ‚‚:', style=style, layout=layout)
        self.w1 = widgets.FloatSlider(value=0.5, min=-2, max=2, step=0.1, description='wâ‚:', style=style, layout=layout)
        self.w2 = widgets.FloatSlider(value=1.2, min=-2, max=2, step=0.1, description='wâ‚‚:', style=style, layout=layout)
        self.bias = widgets.FloatSlider(value=-0.5, min=-3, max=3, step=0.1, description='b:', style=style, layout=layout)
        
        self.activation_type = widgets.RadioButtons(
            options=['Step', 'Sigmoid', 'ReLU'], value='Sigmoid',
            layout=widgets.Layout(width='120px')
        )
        
        controls = widgets.HBox([
            widgets.VBox([widgets.HTML('<b style="color:#4ECDC4">Inputs</b>'), self.x1, self.x2]),
            widgets.VBox([widgets.HTML('<b style="color:#FF6B6B">Weights</b>'), self.w1, self.w2]),
            widgets.VBox([widgets.HTML('<b style="color:#f59e0b">Bias</b>'), self.bias]),
            widgets.VBox([widgets.HTML('<b style="color:#8b5cf6">Activation</b>'), self.activation_type])
        ], layout=widgets.Layout(justify_content='center', gap='20px'))
        
        self.output = widgets.Output()
        
        for w in [self.x1, self.x2, self.w1, self.w2, self.bias]:
            w.observe(self.update, 'value')
        self.activation_type.observe(self.update, 'value')
        
        display(controls, self.output)
        self.update(None)
    
    def update(self, _):
        x = [self.x1.value, self.x2.value]
        w = [self.w1.value, self.w2.value]
        b = self.bias.value
        
        weighted = [x[i] * w[i] for i in range(2)]
        total = sum(weighted) + b
        
        act = self.activation_type.value
        if act == 'Step':
            out = 1.0 if total >= 0 else 0.0
        elif act == 'Sigmoid':
            out = self.sigmoid(total)
        else:
            out = max(0, total)
        
        with self.output:
            clear_output(wait=True)
            
            fig = make_subplots(rows=1, cols=2, column_widths=[0.6, 0.4],
                               subplot_titles=('<b>Computation Flow</b>', f'<b>{act} Function</b>'))
            
            # Pipeline
            stages = [0, 1.3, 2.6, 3.6, 4.6, 5.8]
            
            for i, (xi, wi) in enumerate(zip(x, w)):
                y = 2.5 - i
                col = '#4ECDC4' if i == 0 else '#45B7D1'
                wcol = '#22c55e' if wi >= 0 else '#ef4444'
                
                fig.add_trace(go.Scatter(x=[stages[0]], y=[y], mode='markers+text',
                    marker=dict(size=40, color=col), text=[f'{xi:.1f}'],
                    textposition='middle center', textfont=dict(color='white', size=11),
                    showlegend=False), row=1, col=1)
                
                fig.add_trace(go.Scatter(x=[stages[1]], y=[y], mode='markers+text',
                    marker=dict(size=35, color=wcol, symbol='square'), text=[f'{weighted[i]:.2f}'],
                    textposition='middle center', textfont=dict(color='white', size=10),
                    showlegend=False), row=1, col=1)
                
                fig.add_annotation(x=stages[1]-0.25, y=y, ax=stages[0]+0.2, ay=y,
                    showarrow=True, arrowhead=2, arrowcolor=wcol)
                fig.add_annotation(x=(stages[0]+stages[1])/2, y=y+0.25, text=f'Ã—{wi:.1f}',
                    showarrow=False, font=dict(size=9, color=wcol))
                fig.add_annotation(x=stages[2]-0.25, y=2, ax=stages[1]+0.2, ay=y,
                    showarrow=True, arrowhead=2, arrowcolor='gray')
            
            # Sum
            fig.add_trace(go.Scatter(x=[stages[2]], y=[2], mode='markers+text',
                marker=dict(size=45, color='#667eea'), text=['Î£'],
                textposition='middle center', textfont=dict(color='white', size=18),
                showlegend=False), row=1, col=1)
            
            # Bias box
            fig.add_shape(type='rect', x0=stages[3]-0.3, x1=stages[3]+0.3, y0=1.65, y1=2.35,
                         fillcolor='rgba(245,158,11,0.9)', line=dict(color='#d97706', width=2), row=1, col=1)
            fig.add_annotation(x=stages[3], y=2.1, text='<b>+b</b>', showarrow=False,
                             font=dict(color='white', size=12))
            fig.add_annotation(x=stages[3], y=1.85, text=f'{b:.1f}', showarrow=False,
                             font=dict(color='white', size=10))
            
            # Activation box
            fig.add_shape(type='rect', x0=stages[4]-0.35, x1=stages[4]+0.35, y0=1.6, y1=2.4,
                         fillcolor='rgba(139,92,246,0.9)', line=dict(color='#7c3aed', width=2), row=1, col=1)
            fig.add_annotation(x=stages[4], y=2.1, text=f'<b>{act[0]}(z)</b>', showarrow=False,
                             font=dict(color='white', size=12))
            
            # Output
            ocol = '#22c55e' if out > 0.5 else '#94a3b8'
            fig.add_trace(go.Scatter(x=[stages[5]], y=[2], mode='markers+text',
                marker=dict(size=45, color=ocol), text=[f'{out:.2f}'],
                textposition='middle center', textfont=dict(color='white', size=12),
                showlegend=False), row=1, col=1)
            
            # Arrows
            for i in range(2, 5):
                fig.add_annotation(x=stages[i+1]-0.3, y=2, ax=stages[i]+0.3, ay=2,
                    showarrow=True, arrowhead=2, arrowcolor='gray')
            
            # Activation curve
            z = np.linspace(-5, 5, 100)
            if act == 'Step':
                a = np.where(z >= 0, 1, 0)
            elif act == 'Sigmoid':
                a = self.sigmoid(z)
            else:
                a = np.maximum(0, z)
            
            fig.add_trace(go.Scatter(x=z, y=a, mode='lines', line=dict(color='#8b5cf6', width=3),
                                    showlegend=False), row=1, col=2)
            fig.add_trace(go.Scatter(x=[total], y=[out], mode='markers',
                marker=dict(size=15, color='#ef4444', symbol='star'), showlegend=False), row=1, col=2)
            
            fig.update_layout(height=380, width=1000, showlegend=False, margin=dict(l=30, r=30, t=50, b=30))
            fig.update_xaxes(visible=False, range=[-0.3, 6.3], row=1, col=1)
            fig.update_yaxes(visible=False, range=[1, 3.2], row=1, col=1)
            fig.update_xaxes(title='z', row=1, col=2)
            fig.update_yaxes(title='output', row=1, col=2)
            
            fig.show()

# --- Cross-platform rendering ------------------------------------------------
# Colab cannot render Plotly inside an ipywidgets Output, but it DOES run
# HTML + JS in its sandboxed output frame. JupyterLab/Notebook strip <script>
# from HTML output, but ipywidgets + Plotly works there. So pick per platform.
def _in_colab():
    try:
        import google.colab  # noqa: F401
        return True
    except ImportError:
        return False

_PERCEPTRON_HTML = """<div style="font-family:'Inter',system-ui,sans-serif;background:#ffffff;color:#1f2430;border-radius:14px;padding:18px;box-shadow:0 6px 24px rgba(102,126,234,0.18);max-width:1040px;margin:0 auto;">
  <div style="background:linear-gradient(135deg,#667eea,#764ba2);color:#fff;padding:14px;border-radius:12px;text-align:center;margin-bottom:16px;">
    <h3 style="margin:0;">&#128295; Perceptron Step-by-Step</h3>
  </div>
  <div style="display:flex;justify-content:center;gap:34px;flex-wrap:wrap;margin-bottom:4px;">
    <div>
      <div style="font-weight:700;color:#15a3a3;margin-bottom:6px;">Inputs</div>
      <div>x&#8321; <input id="ppn_x1" type="range" min="0" max="1" step="0.1" value="0.6" style="width:120px;vertical-align:middle;"> <b id="ppn_x1v">0.60</b></div>
      <div>x&#8322; <input id="ppn_x2" type="range" min="0" max="1" step="0.1" value="0.4" style="width:120px;vertical-align:middle;"> <b id="ppn_x2v">0.40</b></div>
    </div>
    <div>
      <div style="font-weight:700;color:#e0484b;margin-bottom:6px;">Weights</div>
      <div>w&#8321; <input id="ppn_w1" type="range" min="-2" max="2" step="0.1" value="0.5" style="width:120px;vertical-align:middle;"> <b id="ppn_w1v">0.50</b></div>
      <div>w&#8322; <input id="ppn_w2" type="range" min="-2" max="2" step="0.1" value="1.2" style="width:120px;vertical-align:middle;"> <b id="ppn_w2v">1.20</b></div>
    </div>
    <div>
      <div style="font-weight:700;color:#d97706;margin-bottom:6px;">Bias</div>
      <div>b <input id="ppn_b" type="range" min="-3" max="3" step="0.1" value="-0.5" style="width:120px;vertical-align:middle;"> <b id="ppn_bv">-0.50</b></div>
    </div>
    <div>
      <div style="font-weight:700;color:#7c3aed;margin-bottom:6px;">Activation</div>
      <label style="display:block;"><input type="radio" name="ppn_act" value="Step"> Step</label>
      <label style="display:block;"><input type="radio" name="ppn_act" value="Sigmoid" checked> Sigmoid</label>
      <label style="display:block;"><input type="radio" name="ppn_act" value="ReLU"> ReLU</label>
    </div>
  </div>
  <div id="ppn_plot" style="width:100%;height:380px;"></div>
</div>
<script>
(function(){
  function boot(){
    var gd = document.getElementById('ppn_plot');
    if(!gd || !window.Plotly){ return; }
    function g(id){ return document.getElementById(id); }
    function sig(x){ x = Math.max(-500, Math.min(500, x)); return 1/(1+Math.exp(-x)); }
    function actVal(name, z){ if(name==='Step') return z>=0?1:0; if(name==='Sigmoid') return sig(z); return Math.max(0,z); }
    function render(){
      var x1=+g('ppn_x1').value, x2=+g('ppn_x2').value;
      var w1=+g('ppn_w1').value, w2=+g('ppn_w2').value, b=+g('ppn_b').value;
      var act = document.querySelector('input[name="ppn_act"]:checked').value;
      g('ppn_x1v').textContent=x1.toFixed(2); g('ppn_x2v').textContent=x2.toFixed(2);
      g('ppn_w1v').textContent=w1.toFixed(2); g('ppn_w2v').textContent=w2.toFixed(2);
      g('ppn_bv').textContent=b.toFixed(2);
      var x=[x1,x2], w=[w1,w2], wt=[x1*w1, x2*w2];
      var total = wt[0]+wt[1]+b, out = actVal(act, total);
      var stages=[0,1.3,2.6,3.6,4.6,5.8];
      var data=[], shapes=[], ann=[];
      for(var i=0;i<2;i++){
        var y=2.5-i;
        var col = i===0 ? '#1bb7b7' : '#3aa0c4';
        var wcol = w[i]>=0 ? '#22c55e' : '#ef4444';
        data.push({x:[stages[0]], y:[y], type:'scatter', mode:'markers+text', xaxis:'x', yaxis:'y',
          marker:{size:40, color:col}, text:[x[i].toFixed(1)], textposition:'middle center',
          textfont:{color:'#fff', size:11}, hoverinfo:'skip'});
        data.push({x:[stages[1]], y:[y], type:'scatter', mode:'markers+text', xaxis:'x', yaxis:'y',
          marker:{size:34, color:wcol, symbol:'square'}, text:[wt[i].toFixed(2)], textposition:'middle center',
          textfont:{color:'#fff', size:10}, hoverinfo:'skip'});
        ann.push({x:stages[1]-0.25, y:y, ax:stages[0]+0.2, ay:y, xref:'x', yref:'y', axref:'x', ayref:'y',
          showarrow:true, arrowhead:2, arrowwidth:1.4, arrowcolor:wcol});
        ann.push({x:(stages[0]+stages[1])/2, y:y+0.28, text:'&#215;'+w[i].toFixed(1), showarrow:false,
          xref:'x', yref:'y', font:{size:10, color:wcol}});
        ann.push({x:stages[2]-0.28, y:2, ax:stages[1]+0.2, ay:y, xref:'x', yref:'y', axref:'x', ayref:'y',
          showarrow:true, arrowhead:2, arrowwidth:1.2, arrowcolor:'#94a3b8'});
      }
      data.push({x:[stages[2]], y:[2], type:'scatter', mode:'markers+text', xaxis:'x', yaxis:'y',
        marker:{size:46, color:'#667eea'}, text:['\u03A3'], textposition:'middle center',
        textfont:{color:'#fff', size:18}, hoverinfo:'skip'});
      shapes.push({type:'rect', xref:'x', yref:'y', x0:stages[3]-0.3, x1:stages[3]+0.3, y0:1.65, y1:2.35,
        fillcolor:'rgba(245,158,11,0.92)', line:{color:'#d97706', width:2}});
      ann.push({x:stages[3], y:2.12, text:'<b>+b</b>', showarrow:false, xref:'x', yref:'y', font:{color:'#fff', size:12}});
      ann.push({x:stages[3], y:1.86, text:b.toFixed(1), showarrow:false, xref:'x', yref:'y', font:{color:'#fff', size:10}});
      shapes.push({type:'rect', xref:'x', yref:'y', x0:stages[4]-0.35, x1:stages[4]+0.35, y0:1.6, y1:2.4,
        fillcolor:'rgba(139,92,246,0.92)', line:{color:'#7c3aed', width:2}});
      ann.push({x:stages[4], y:2.1, text:'<b>'+act.charAt(0)+'(z)</b>', showarrow:false, xref:'x', yref:'y', font:{color:'#fff', size:12}});
      var ocol = out>0.5 ? '#22c55e' : '#94a3b8';
      data.push({x:[stages[5]], y:[2], type:'scatter', mode:'markers+text', xaxis:'x', yaxis:'y',
        marker:{size:46, color:ocol}, text:[out.toFixed(2)], textposition:'middle center',
        textfont:{color:'#fff', size:12}, hoverinfo:'skip'});
      for(var s=2;s<5;s++){
        ann.push({x:stages[s+1]-0.3, y:2, ax:stages[s]+0.3, ay:2, xref:'x', yref:'y', axref:'x', ayref:'y',
          showarrow:true, arrowhead:2, arrowwidth:1.2, arrowcolor:'#94a3b8'});
      }
      var zs=[], as_=[];
      for(var k=0;k<120;k++){ var z=-5+10*k/119; zs.push(z); as_.push(actVal(act, z)); }
      data.push({x:zs, y:as_, type:'scatter', mode:'lines', xaxis:'x2', yaxis:'y2',
        line:{color:'#8b5cf6', width:3}, hoverinfo:'skip'});
      data.push({x:[total], y:[out], type:'scatter', mode:'markers', xaxis:'x2', yaxis:'y2',
        marker:{size:15, color:'#ef4444', symbol:'star'}, hoverinfo:'skip'});
      var titleAnn=[
        {x:0.27, y:1.08, xref:'paper', yref:'paper', text:'<b>Computation Flow</b>', showarrow:false, font:{size:13, color:'#444'}},
        {x:0.85, y:1.08, xref:'paper', yref:'paper', text:'<b>'+act+' Function</b>', showarrow:false, font:{size:13, color:'#444'}}
      ];
      var layout={
        height:380, margin:{l:30, r:30, t:46, b:34}, showlegend:false,
        paper_bgcolor:'rgba(0,0,0,0)', plot_bgcolor:'rgba(0,0,0,0)',
        xaxis:{domain:[0,0.56], visible:false, range:[-0.3,6.3], anchor:'y'},
        yaxis:{domain:[0,1], visible:false, range:[1,3.2], anchor:'x'},
        xaxis2:{domain:[0.66,1], anchor:'y2', title:{text:'z'}, color:'#555', zeroline:false},
        yaxis2:{anchor:'x2', title:{text:'output'}, color:'#555', zeroline:false},
        shapes:shapes, annotations:ann.concat(titleAnn)
      };
      Plotly.react(gd, data, layout, {displayModeBar:false, responsive:true});
    }
    ['ppn_x1','ppn_x2','ppn_w1','ppn_w2','ppn_b'].forEach(function(id){
      g(id).addEventListener('input', render);
    });
    document.querySelectorAll('input[name="ppn_act"]').forEach(function(r){ r.addEventListener('change', render); });
    render();
  }
  if(window.Plotly){ boot(); }
  else {
    var ex = document.getElementById('ppn_plotly_cdn');
    if(ex){ ex.addEventListener('load', boot); }
    else {
      var sc = document.createElement('script');
      sc.id = 'ppn_plotly_cdn';
      sc.src = 'https://cdn.plot.ly/plotly-2.35.2.min.js';
      sc.onload = boot;
      document.head.appendChild(sc);
    }
  }
})();
</script>
"""

if _in_colab():
    display(HTML(_PERCEPTRON_HTML))
else:
    PerceptronViz()
```

```python
# =============================================================================
# PART 3: WATCH A NETWORK LEARN (PRE-COMPUTED VISUALIZATION)
# =============================================================================

display(HTML("""
<div class="concept-box">
    <h2 style="margin-top:0; color:#667eea;">ðŸŽ¬ Watch a Deep Network Learn!</h2>
    <p>Use the slider to see how neurons <strong>activate and change</strong> as the network trains.</p>
    <p>Notice how early training shows random patterns, then meaningful structure emerges!</p>
</div>
"""))

# Native Plotly animation (frames + slider + play button).
# Renders in Colab, JupyterLab, Notebook 7, classic notebook, and static HTML.

import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ---- Network structure: 3 -> 5 -> 4 -> 2 ----
layers = [3, 5, 4, 2]
colors = ['#00ffff', '#00ddff', '#00bbff', '#00ff88']
layer_names = ['Input', 'Hidden 1', 'Hidden 2', 'Output']

# Node positions (3D)
positions = []
for li, n in enumerate(layers):
    layer_pos = []
    for i in range(n):
        z = (i - (n - 1) / 2) * 1.0
        layer_pos.append((li * 2.5, 0, z))
    positions.append(layer_pos)

# Pre-compute activation patterns per epoch-frame
n_frames = 50
np.random.seed(42)
activation_frames = []
for fr in range(n_frames):
    t = fr / (n_frames - 1)
    fa = []
    for li, n in enumerate(layers):
        if fr == 0:
            acts = np.random.uniform(0.2, 0.8, n)
        else:
            base = np.random.uniform(0.3, 0.7, n)
            structure = np.sin(np.linspace(0, np.pi, n) + li) * 0.3
            acts = base * (1 - t) + (0.5 + structure * t)
            acts += np.sin(fr * 0.3 + np.arange(n) * 0.5) * 0.1 * (1 - t * 0.5)
        acts = np.clip(acts, 0.1, 0.95)
        fa.append(acts)
    activation_frames.append(fa)

# Pre-compute loss curve
np.random.seed(7)
xs_loss = np.linspace(0, 1, n_frames)
loss_curve = 0.9 * np.exp(-3 * xs_loss) + 0.1 + np.random.randn(n_frames) * 0.02 * (1 - xs_loss)
loss_curve = np.clip(loss_curve, 0.08, 1.0)

# ---- Static edges ----
edge_x, edge_y, edge_z = [], [], []
for li in range(len(layers) - 1):
    for (x1, y1, z1) in positions[li]:
        for (x2, y2, z2) in positions[li + 1]:
            edge_x += [x1, x2, None]
            edge_y += [y1, y2, None]
            edge_z += [z1, z2, None]


def node_traces(acts):
    """Return [glow0, main0, glow1, main1, ...] for the 4 layers."""
    out = []
    for li, (pos, a) in enumerate(zip(positions, acts)):
        xs = [p[0] for p in pos]
        ys = [p[1] for p in pos]
        zs = [p[2] for p in pos]
        sizes = (10 + 20 * np.array(a))
        out.append(go.Scatter3d(
            x=xs, y=ys, z=zs, mode='markers',
            marker=dict(size=(sizes * 2).tolist(), color=colors[li], opacity=0.15),
            hoverinfo='skip', showlegend=False, scene='scene'))
        out.append(go.Scatter3d(
            x=xs, y=ys, z=zs, mode='markers',
            marker=dict(size=sizes.tolist(), color=colors[li], opacity=0.9,
                        line=dict(color='white', width=0.5)),
            text=[f'{layer_names[li]}<br>Activation: {a[i]:.2f}' for i in range(len(a))],
            hoverinfo='text', showlegend=False, scene='scene'))
    return out


def loss_traces(fr):
    line = go.Scatter(
        x=list(range(fr + 1)), y=loss_curve[:fr + 1].tolist(),
        xaxis='x2', yaxis='y2', mode='lines',
        line=dict(color='#00ccff', width=3), fill='tozeroy',
        fillcolor='rgba(0,204,255,0.2)', showlegend=False)
    pt = go.Scatter(
        x=[fr], y=[float(loss_curve[fr])], xaxis='x2', yaxis='y2', mode='markers',
        marker=dict(size=12, color='#ff4444'), showlegend=False)
    return [line, pt]


def metric_annotations(fr):
    epoch = fr * 10
    l = float(loss_curve[fr])
    acc = min(99.0, max(50.0, 50 + 50 * (1 - l) * 1.1))
    return [
        dict(x=0.16, y=1.04, xref='paper', yref='paper', showarrow=False, align='center',
             text=f'<b>Epoch</b>  <span style="font-size:20px">{epoch}</span>',
             font=dict(color='#aab4d4', size=13)),
        dict(x=0.34, y=1.04, xref='paper', yref='paper', showarrow=False, align='center',
             text=f'<b>Loss</b>  <span style="font-size:20px;color:#ff8a8a">{l:.3f}</span>',
             font=dict(color='#aab4d4', size=13)),
        dict(x=0.54, y=1.04, xref='paper', yref='paper', showarrow=False, align='center',
             text=f'<b>Accuracy</b>  <span style="font-size:20px;color:#7CFC9A">{acc:.1f}%</span>',
             font=dict(color='#aab4d4', size=13)),
    ]


# ---- Base figure (frame 0) ----
fig = make_subplots(rows=1, cols=2,
                    specs=[[{"type": "scatter3d"}, {"type": "xy"}]],
                    column_widths=[0.7, 0.3], horizontal_spacing=0.06)

# index 0: edges
fig.add_trace(go.Scatter3d(x=edge_x, y=edge_y, z=edge_z, mode='lines',
                           line=dict(color='rgba(0,180,255,0.25)', width=1.5),
                           hoverinfo='skip', showlegend=False), row=1, col=1)
# index 1..8: node glow/main for frame 0
for tr in node_traces(activation_frames[0]):
    fig.add_trace(tr, row=1, col=1)
# index 9..12: static labels
for li, name in enumerate(layer_names):
    fig.add_trace(go.Scatter3d(x=[li * 2.5], y=[0], z=[-3.2], mode='text',
                               text=[f'<b>{name}</b>'],
                               textfont=dict(size=13, color=colors[li]),
                               hoverinfo='skip', showlegend=False), row=1, col=1)
# index 13,14: loss line + point for frame 0
for tr in loss_traces(0):
    fig.add_trace(tr, row=1, col=2)

ANIM_IDX = [1, 2, 3, 4, 5, 6, 7, 8, 13, 14]

# ---- Frames ----
frames = []
for fr in range(n_frames):
    fdata = node_traces(activation_frames[fr]) + loss_traces(fr)
    frames.append(go.Frame(name=str(fr), data=fdata, traces=ANIM_IDX,
                           layout=go.Layout(annotations=metric_annotations(fr))))
fig.frames = frames

# ---- Slider + play/pause ----
slider_steps = []
for fr in range(n_frames):
    slider_steps.append(dict(
        method='animate', label=str(fr * 10),
        args=[[str(fr)], dict(mode='immediate',
                              frame=dict(duration=0, redraw=True),
                              transition=dict(duration=0))]))

fig.update_layout(
    height=600, showlegend=False,
    paper_bgcolor='rgb(10,15,30)', plot_bgcolor='rgb(10,15,30)',
    margin=dict(l=0, r=10, t=70, b=70),
    annotations=metric_annotations(0),
    scene=dict(
        domain=dict(x=[0.0, 0.64], y=[0.0, 1.0]),
        xaxis=dict(showgrid=True, gridcolor='rgba(100,150,200,0.1)', showbackground=True,
                   backgroundcolor='rgb(10,15,30)', showticklabels=False, title=''),
        yaxis=dict(showgrid=True, gridcolor='rgba(100,150,200,0.1)', showbackground=True,
                   backgroundcolor='rgb(10,15,30)', showticklabels=False, title=''),
        zaxis=dict(showgrid=True, gridcolor='rgba(100,150,200,0.1)', showbackground=True,
                   backgroundcolor='rgb(10,15,30)', showticklabels=False, title=''),
        camera=dict(eye=dict(x=1.3, y=1.8, z=0.6)),
        aspectmode='manual', aspectratio=dict(x=2, y=0.5, z=1.2)),
    updatemenus=[dict(
        type='buttons', direction='left', showactive=False,
        x=0.0, y=-0.04, xanchor='left', yanchor='top',
        bgcolor='rgba(102,126,234,0.25)', bordercolor='rgba(102,126,234,0.6)',
        font=dict(color='#dfe6ff'),
        buttons=[
            dict(label='▶ Play', method='animate',
                 args=[None, dict(frame=dict(duration=120, redraw=True),
                                  fromcurrent=True, transition=dict(duration=0))]),
            dict(label='⏸ Pause', method='animate',
                 args=[[None], dict(mode='immediate',
                                    frame=dict(duration=0, redraw=True),
                                    transition=dict(duration=0))])])],
    sliders=[dict(active=0, x=0.08, y=-0.02, len=0.9,
                  pad=dict(t=10, b=10),
                  currentvalue=dict(prefix='Epoch: ', font=dict(color='#aab4d4', size=13)),
                  font=dict(color='#9aa4c4'),
                  steps=slider_steps)])

fig.update_xaxes(title='Epoch', range=[0, 50], gridcolor='rgba(100,150,200,0.2)',
                 color='#aab4d4', domain=[0.72, 1.0], row=1, col=2)
fig.update_yaxes(title='Loss', range=[0, 1], gridcolor='rgba(100,150,200,0.2)',
                 color='#aab4d4', row=1, col=2)

fig.show()

display(HTML("""
<div class="insight-box">
    <h3 style="margin-top:0; color:#856404;">ðŸ’¡ What You're Seeing</h3>
    <ul style="margin-bottom: 0;">
        <li><strong>Node size & brightness</strong> = activation level (how "active" that neuron is)</li>
        <li><strong>Early training</strong> = chaotic, random activations</li>
        <li><strong>Later training</strong> = organized, meaningful patterns emerge</li>
        <li><strong>Loss decreasing</strong> = network is learning!</li>
    </ul>
    <p style="margin-top: 10px;"><em>Drag the slider or click Play to see the network learn!</em></p>
</div>
"""))
```
