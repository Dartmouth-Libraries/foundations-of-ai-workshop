```python
# =============================================================================
# PART 1: BIOLOGICAL vs ARTIFICIAL NEURON
# =============================================================================
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
import warnings
warnings.filterwarnings('ignore')

import plotly.io as pio
pio.renderers.default = "notebook"
display(HTML("""
<div class="concept-box">
    <h2 style="margin-top:0; color:#667eea;">🔬 The Perceptron: Loosely Inspired by Biology</h2>
    <p>In 1958, Frank Rosenblatt created the <strong>perceptron</strong> — taking inspiration from how biological neurons work.</p>
    <p>Let's compare the two side by side!</p>
</div>
"""))

def create_biological_neuron_detailed():
    """Create a detailed, anatomically-inspired biological neuron visualization"""
    
    fig = go.Figure()
    
    np.random.seed(42)
    dendrite_color = '#8B4513'
    
    def draw_branch(start_x, start_y, angle, length, depth, traces):
        if depth == 0 or length < 0.1:
            return
        
        end_x = start_x + length * np.cos(angle)
        end_y = start_y + length * np.sin(angle)
        
        traces.append(go.Scatter(
            x=[start_x, end_x], y=[start_y, end_y],
            mode='lines',
            line=dict(color=dendrite_color, width=max(1, 4 - depth)),
            hoverinfo='skip', showlegend=False
        ))
        
        if depth >= 2:
            traces.append(go.Scatter(
                x=[end_x], y=[end_y],
                mode='markers',
                marker=dict(size=6, color='#D2691E'),
                hoverinfo='text', hovertext='Synapse',
                showlegend=False
            ))
        
        if depth > 0:
            for _ in range(np.random.randint(2, 4)):
                new_angle = angle + np.random.uniform(-0.7, 0.7)
                new_length = length * np.random.uniform(0.5, 0.8)
                draw_branch(end_x, end_y, new_angle, new_length, depth - 1, traces)
    
    dendrite_traces = []
    soma_x, soma_y = 2.5, 2.5
    
    for angle in [2.5, 2.8, 3.1, 3.5, 3.9, 2.2]:
        draw_branch(soma_x, soma_y, angle, 1.2, 3, dendrite_traces)
    
    for trace in dendrite_traces:
        fig.add_trace(trace)
    
    # Soma
    theta = np.linspace(0, 2*np.pi, 100)
    r = 0.5 + 0.08 * np.sin(5*theta) + 0.05 * np.sin(8*theta)
    fig.add_trace(go.Scatter(
        x=soma_x + r * np.cos(theta), y=soma_y + r * np.sin(theta),
        fill='toself', fillcolor='rgba(255, 182, 193, 0.9)',
        line=dict(color='#FF69B4', width=3),
        hoverinfo='text', hovertext='Cell Body (Soma)', showlegend=False
    ))
    
    # Nucleus
    fig.add_trace(go.Scatter(
        x=soma_x + 0.25 * np.cos(theta), y=soma_y + 0.25 * np.sin(theta),
        fill='toself', fillcolor='rgba(138, 43, 226, 0.7)',
        line=dict(color='#9400D3', width=2),
        hoverinfo='text', hovertext='Nucleus', showlegend=False
    ))
    
    # Axon
    axon_x = np.linspace(soma_x + 0.8, soma_x + 4, 50)
    axon_y = soma_y - 0.5 + 0.15 * np.sin(3 * (axon_x - soma_x))
    fig.add_trace(go.Scatter(x=axon_x, y=axon_y, mode='lines',
                             line=dict(color='#4169E1', width=6), showlegend=False))
    
    # Myelin
    for mx in np.linspace(soma_x + 1.2, soma_x + 3.5, 5):
        my = soma_y - 0.5 + 0.15 * np.sin(3 * (mx - soma_x))
        fig.add_trace(go.Scatter(
            x=mx + 0.2 * np.cos(theta), y=my + 0.12 * np.sin(theta),
            fill='toself', fillcolor='rgba(255, 215, 0, 0.6)',
            line=dict(color='#FFD700', width=1), showlegend=False, hoverinfo='skip'
        ))
    
    # Terminals
    term_x = soma_x + 4
    term_y = soma_y - 0.5 + 0.15 * np.sin(3 * (term_x - soma_x))
    for tx, ty in [(term_x + 0.6, term_y + 0.3), (term_x + 0.7, term_y), (term_x + 0.6, term_y - 0.3)]:
        fig.add_trace(go.Scatter(x=[term_x, tx], y=[term_y, ty], mode='lines',
                                 line=dict(color='#4169E1', width=3), showlegend=False))
        fig.add_trace(go.Scatter(x=[tx], y=[ty], mode='markers',
                                 marker=dict(size=15, color='#FF4500'), showlegend=False))
    
    # Labels
    for lx, ly, text, color in [(1.0, 4.2, 'Dendrites', '#8B4513'), (soma_x, soma_y+0.9, 'Soma', '#FF69B4'),
                                 (soma_x+2.5, soma_y+0.3, 'Axon', '#4169E1'), (soma_x+4.8, term_y, 'Terminals', '#FF4500')]:
        fig.add_annotation(x=lx, y=ly, text=f'<b>{text}</b>', showarrow=False,
                          font=dict(size=11, color=color), bgcolor='rgba(255,255,255,0.8)')
    
    fig.update_layout(height=450, width=700, showlegend=False,
                      xaxis=dict(visible=False, range=[0, 8]),
                      yaxis=dict(visible=False, range=[0, 5], scaleanchor='x'),
                      margin=dict(l=20, r=20, t=40, b=20))
    return fig

fig_bio = create_biological_neuron_detailed()
fig_bio.show()

display(HTML("""
<div class="insight-box">
    <h3 style="margin-top:0; color:#856404;">💡 How a Biological Neuron Works</h3>
    <ol style="margin-bottom: 0;">
        <li><strong>Dendrites</strong> receive <em>electrochemical signals</em> from thousands of other neurons</li>
        <li><strong>Soma</strong> integrates all incoming signals</li>
        <li>If the total exceeds a <strong>threshold</strong>, the neuron "fires"</li>
        <li>Signal travels down the <strong>axon</strong> to the terminals</li>
    </ol>
</div>
"""))
```

```python
# =============================================================================
# PART 2: THE ARTIFICIAL NEURON - STEP BY STEP
# =============================================================================

display(HTML("""
<div class="concept-box">
    <h2 style="margin-top:0; color:#667eea;">🤖 The Artificial Neuron (Perceptron)</h2>
    <p>Now let's see the artificial version - broken into <strong>clear, separate steps</strong>!</p>
</div>
"""))

class PerceptronViz:
    def __init__(self):
        self.create_interface()
    
    def sigmoid(self, x):
        return 1 / (1 + np.exp(-np.clip(x, -500, 500)))
    
    def create_interface(self):
        display(HTML("""
        <div style="background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); 
                    color: white; padding: 20px; border-radius: 12px; text-align: center;
                    margin-bottom: 20px;">
            <h3 style="margin: 0;">🔧 Perceptron Step-by-Step</h3>
        </div>
        """))
        
        layout = widgets.Layout(width='180px')
        style = {'description_width': '40px'}
        
        self.x1 = widgets.FloatSlider(value=0.6, min=0, max=1, step=0.1, description='x₁:', style=style, layout=layout)
        self.x2 = widgets.FloatSlider(value=0.4, min=0, max=1, step=0.1, description='x₂:', style=style, layout=layout)
        self.w1 = widgets.FloatSlider(value=0.5, min=-2, max=2, step=0.1, description='w₁:', style=style, layout=layout)
        self.w2 = widgets.FloatSlider(value=1.2, min=-2, max=2, step=0.1, description='w₂:', style=style, layout=layout)
        self.bias = widgets.FloatSlider(value=-0.5, min=-3, max=3, step=0.1, description='b:', style=style, layout=layout)
        
        self.activation_type = widgets.RadioButtons(
            options=['Step', 'Sigmoid', 'ReLU'], value='Sigmoid',
            layout=widgets.Layout(width='120px')
        )
        
        controls = widgets.HBox([
            widgets.VBox([widgets.HTML('<b style="color:#4ECDC4">Inputs</b>'), self.x1, self.x2]),
            widgets.VBox([widgets.HTML('<b style="color:#FF6B6B">Weights</b>'), self.w1, self.w2]),
            widgets.VBox([widgets.HTML('<b style="color:#f59e0b">Bias</b>'), self.bias]),
            widgets.VBox([widgets.HTML('<b style="color:#8b5cf6">Activation</b>'), self.activation_type])
        ], layout=widgets.Layout(justify_content='center', gap='20px'))
        
        self.output = widgets.Output()
        
        for w in [self.x1, self.x2, self.w1, self.w2, self.bias]:
            w.observe(self.update, 'value')
        self.activation_type.observe(self.update, 'value')
        
        display(controls, self.output)
        self.update(None)
    
    def update(self, _):
        x = [self.x1.value, self.x2.value]
        w = [self.w1.value, self.w2.value]
        b = self.bias.value
        
        weighted = [x[i] * w[i] for i in range(2)]
        total = sum(weighted) + b
        
        act = self.activation_type.value
        if act == 'Step':
            out = 1.0 if total >= 0 else 0.0
        elif act == 'Sigmoid':
            out = self.sigmoid(total)
        else:
            out = max(0, total)
        
        with self.output:
            clear_output(wait=True)
            
            fig = make_subplots(rows=1, cols=2, column_widths=[0.6, 0.4],
                               subplot_titles=('<b>Computation Flow</b>', f'<b>{act} Function</b>'))
            
            # Pipeline
            stages = [0, 1.3, 2.6, 3.6, 4.6, 5.8]
            
            for i, (xi, wi) in enumerate(zip(x, w)):
                y = 2.5 - i
                col = '#4ECDC4' if i == 0 else '#45B7D1'
                wcol = '#22c55e' if wi >= 0 else '#ef4444'
                
                fig.add_trace(go.Scatter(x=[stages[0]], y=[y], mode='markers+text',
                    marker=dict(size=40, color=col), text=[f'{xi:.1f}'],
                    textposition='middle center', textfont=dict(color='white', size=11),
                    showlegend=False), row=1, col=1)
                
                fig.add_trace(go.Scatter(x=[stages[1]], y=[y], mode='markers+text',
                    marker=dict(size=35, color=wcol, symbol='square'), text=[f'{weighted[i]:.2f}'],
                    textposition='middle center', textfont=dict(color='white', size=10),
                    showlegend=False), row=1, col=1)
                
                fig.add_annotation(x=stages[1]-0.25, y=y, ax=stages[0]+0.2, ay=y,
                    showarrow=True, arrowhead=2, arrowcolor=wcol)
                fig.add_annotation(x=(stages[0]+stages[1])/2, y=y+0.25, text=f'×{wi:.1f}',
                    showarrow=False, font=dict(size=9, color=wcol))
                fig.add_annotation(x=stages[2]-0.25, y=2, ax=stages[1]+0.2, ay=y,
                    showarrow=True, arrowhead=2, arrowcolor='gray')
            
            # Sum
            fig.add_trace(go.Scatter(x=[stages[2]], y=[2], mode='markers+text',
                marker=dict(size=45, color='#667eea'), text=['Σ'],
                textposition='middle center', textfont=dict(color='white', size=18),
                showlegend=False), row=1, col=1)
            
            # Bias box
            fig.add_shape(type='rect', x0=stages[3]-0.3, x1=stages[3]+0.3, y0=1.65, y1=2.35,
                         fillcolor='rgba(245,158,11,0.9)', line=dict(color='#d97706', width=2), row=1, col=1)
            fig.add_annotation(x=stages[3], y=2.1, text='<b>+b</b>', showarrow=False,
                             font=dict(color='white', size=12))
            fig.add_annotation(x=stages[3], y=1.85, text=f'{b:.1f}', showarrow=False,
                             font=dict(color='white', size=10))
            
            # Activation box
            fig.add_shape(type='rect', x0=stages[4]-0.35, x1=stages[4]+0.35, y0=1.6, y1=2.4,
                         fillcolor='rgba(139,92,246,0.9)', line=dict(color='#7c3aed', width=2), row=1, col=1)
            fig.add_annotation(x=stages[4], y=2.1, text=f'<b>{act[0]}(z)</b>', showarrow=False,
                             font=dict(color='white', size=12))
            
            # Output
            ocol = '#22c55e' if out > 0.5 else '#94a3b8'
            fig.add_trace(go.Scatter(x=[stages[5]], y=[2], mode='markers+text',
                marker=dict(size=45, color=ocol), text=[f'{out:.2f}'],
                textposition='middle center', textfont=dict(color='white', size=12),
                showlegend=False), row=1, col=1)
            
            # Arrows
            for i in range(2, 5):
                fig.add_annotation(x=stages[i+1]-0.3, y=2, ax=stages[i]+0.3, ay=2,
                    showarrow=True, arrowhead=2, arrowcolor='gray')
            
            # Activation curve
            z = np.linspace(-5, 5, 100)
            if act == 'Step':
                a = np.where(z >= 0, 1, 0)
            elif act == 'Sigmoid':
                a = self.sigmoid(z)
            else:
                a = np.maximum(0, z)
            
            fig.add_trace(go.Scatter(x=z, y=a, mode='lines', line=dict(color='#8b5cf6', width=3),
                                    showlegend=False), row=1, col=2)
            fig.add_trace(go.Scatter(x=[total], y=[out], mode='markers',
                marker=dict(size=15, color='#ef4444', symbol='star'), showlegend=False), row=1, col=2)
            
            fig.update_layout(height=380, width=1000, showlegend=False, margin=dict(l=30, r=30, t=50, b=30))
            fig.update_xaxes(visible=False, range=[-0.3, 6.3], row=1, col=1)
            fig.update_yaxes(visible=False, range=[1, 3.2], row=1, col=1)
            fig.update_xaxes(title='z', row=1, col=2)
            fig.update_yaxes(title='output', row=1, col=2)
            
            fig.show()

PerceptronViz()
```

```python
# =============================================================================
# PART 3: WATCH A NETWORK LEARN (PRE-COMPUTED VISUALIZATION)
# =============================================================================

display(HTML("""
<div class="concept-box">
    <h2 style="margin-top:0; color:#667eea;">🎬 Watch a Deep Network Learn!</h2>
    <p>Use the slider to see how neurons <strong>activate and change</strong> as the network trains.</p>
    <p>Notice how early training shows random patterns, then meaningful structure emerges!</p>
</div>
"""))

class NetworkLearningViz:
    """Pre-computed network visualization - no live training, just beautiful visuals"""
    
    def __init__(self):
        # Network structure: 3 -> 5 -> 4 -> 2
        self.layers = [3, 5, 4, 2]
        
        # Pre-compute node positions
        self.positions = self.compute_positions()
        
        # Pre-compute activation patterns for different "epochs"
        self.activation_frames = self.precompute_activations()
        
        # Pre-compute loss curve
        self.loss_curve = self.precompute_loss()
        
        self.create_interface()
    
    def compute_positions(self):
        """Compute 3D positions for all nodes"""
        positions = []
        for layer_idx, n_neurons in enumerate(self.layers):
            layer_x = layer_idx * 2.5
            layer_pos = []
            for i in range(n_neurons):
                z = (i - (n_neurons - 1) / 2) * 1.0
                layer_pos.append((layer_x, 0, z))
            positions.append(layer_pos)
        return positions
    
    def precompute_activations(self):
        """Pre-compute activation patterns for smooth animation"""
        n_frames = 50
        frames = []
        
        np.random.seed(42)
        
        for frame in range(n_frames):
            t = frame / (n_frames - 1)  # 0 to 1
            
            frame_activations = []
            for layer_idx, n_neurons in enumerate(self.layers):
                if frame == 0:
                    # Random initial state
                    acts = np.random.uniform(0.2, 0.8, n_neurons)
                else:
                    # Gradually become more structured
                    base = np.random.uniform(0.3, 0.7, n_neurons)
                    # Add some "learned" structure
                    structure = np.sin(np.linspace(0, np.pi, n_neurons) + layer_idx) * 0.3
                    acts = base * (1 - t) + (0.5 + structure * t)
                    # Add slight variation
                    acts += np.sin(frame * 0.3 + np.arange(n_neurons) * 0.5) * 0.1 * (1 - t * 0.5)
                
                acts = np.clip(acts, 0.1, 0.95)
                frame_activations.append(acts)
            
            frames.append(frame_activations)
        
        return frames
    
    def precompute_loss(self):
        """Pre-compute a nice loss curve"""
        n_frames = 50
        x = np.linspace(0, 1, n_frames)
        # Exponential decay with some noise
        loss = 0.9 * np.exp(-3 * x) + 0.1 + np.random.randn(n_frames) * 0.02 * (1 - x)
        return np.clip(loss, 0.08, 1.0)
    
    def create_interface(self):
        """Create the slider interface"""
        
        self.epoch_slider = widgets.IntSlider(
            value=0, min=0, max=len(self.activation_frames) - 1, step=1,
            description='Training:',
            style={'description_width': '70px'},
            layout=widgets.Layout(width='500px'),
            continuous_update=True
        )
        
        self.play_btn = widgets.Button(description='▶️ Play', button_style='success',
                                       layout=widgets.Layout(width='100px'))
        self.reset_btn = widgets.Button(description='🔄 Reset', button_style='warning',
                                        layout=widgets.Layout(width='100px'))
        
        self.status_html = widgets.HTML()
        self.plot_output = widgets.Output()
        
        self.epoch_slider.observe(self.update_plot, 'value')
        self.play_btn.on_click(self.play_animation)
        self.reset_btn.on_click(self.reset)
        
        self.is_playing = False
        
        controls = widgets.HBox([self.epoch_slider, self.play_btn, self.reset_btn],
                               layout=widgets.Layout(justify_content='center', gap='15px'))
        
        display(controls)
        display(self.status_html)
        display(self.plot_output)
        
        self.update_plot(None)
    
    def update_plot(self, _):
        """Update the visualization for current epoch"""
        frame = self.epoch_slider.value
        activations = self.activation_frames[frame]
        
        # Compute metrics
        epoch = frame * 10
        loss = self.loss_curve[frame]
        accuracy = 50 + (100 - 50) * (1 - loss) * 1.1
        accuracy = min(99, max(50, accuracy))
        
        # Update status
        self.status_html.value = f"""
        <div style="display: flex; justify-content: center; gap: 25px; margin: 15px 0; font-family: sans-serif;">
            <div style="background: linear-gradient(135deg, #667eea, #764ba2); 
                        color: white; padding: 15px 25px; border-radius: 12px; text-align: center;">
                <div style="font-size: 11px; opacity: 0.9;">Epoch</div>
                <div style="font-size: 28px; font-weight: 700;">{epoch}</div>
            </div>
            <div style="background: linear-gradient(135deg, #FF6B6B, #ee5a5a); 
                        color: white; padding: 15px 25px; border-radius: 12px; text-align: center;">
                <div style="font-size: 11px; opacity: 0.9;">Loss</div>
                <div style="font-size: 28px; font-weight: 700;">{loss:.3f}</div>
            </div>
            <div style="background: linear-gradient(135deg, #22c55e, #16a34a); 
                        color: white; padding: 15px 25px; border-radius: 12px; text-align: center;">
                <div style="font-size: 11px; opacity: 0.9;">Accuracy</div>
                <div style="font-size: 28px; font-weight: 700;">{accuracy:.1f}%</div>
            </div>
        </div>
        """
        
        with self.plot_output:
            clear_output(wait=True)
            
            fig = make_subplots(
                rows=1, cols=2,
                specs=[[{"type": "scatter3d"}, {"type": "scatter"}]],
                column_widths=[0.7, 0.3],
                horizontal_spacing=0.05
            )
            
            # === Build all edges as one trace ===
            edge_x, edge_y, edge_z = [], [], []
            
            for layer_idx in range(len(self.layers) - 1):
                for x1, y1, z1 in self.positions[layer_idx]:
                    for x2, y2, z2 in self.positions[layer_idx + 1]:
                        edge_x.extend([x1, x2, None])
                        edge_y.extend([y1, y2, None])
                        edge_z.extend([z1, z2, None])
            
            fig.add_trace(go.Scatter3d(
                x=edge_x, y=edge_y, z=edge_z,
                mode='lines',
                line=dict(color='rgba(0, 180, 255, 0.25)', width=1.5),
                hoverinfo='skip', showlegend=False
            ), row=1, col=1)
            
            # === Draw nodes with glow ===
            colors = ['#00ffff', '#00ddff', '#00bbff', '#00ff88']
            layer_names = ['Input', 'Hidden 1', 'Hidden 2', 'Output']
            
            for layer_idx, (positions, acts) in enumerate(zip(self.positions, activations)):
                xs, ys, zs = zip(*positions)
                
                # Node sizes pulse with activation
                sizes = 10 + 20 * acts
                
                # Glow (larger, transparent)
                fig.add_trace(go.Scatter3d(
                    x=xs, y=ys, z=zs,
                    mode='markers',
                    marker=dict(size=sizes * 2, color=colors[layer_idx], opacity=0.15),
                    hoverinfo='skip', showlegend=False
                ), row=1, col=1)
                
                # Main nodes
                fig.add_trace(go.Scatter3d(
                    x=xs, y=ys, z=zs,
                    mode='markers',
                    marker=dict(size=sizes, color=colors[layer_idx], opacity=0.9,
                               line=dict(color='white', width=0.5)),
                    text=[f'{layer_names[layer_idx]}<br>Activation: {acts[i]:.2f}' for i in range(len(acts))],
                    hoverinfo='text', showlegend=False
                ), row=1, col=1)
            
            # Layer labels
            for layer_idx, name in enumerate(layer_names):
                fig.add_trace(go.Scatter3d(
                    x=[layer_idx * 2.5], y=[0], z=[-3.2],
                    mode='text', text=[f'<b>{name}</b>'],
                    textfont=dict(size=13, color=colors[layer_idx]),
                    hoverinfo='skip', showlegend=False
                ), row=1, col=1)
            
            # === Loss curve ===
            fig.add_trace(go.Scatter(
                x=list(range(frame + 1)),
                y=self.loss_curve[:frame + 1],
                mode='lines',
                line=dict(color='#00ccff', width=3),
                fill='tozeroy',
                fillcolor='rgba(0, 204, 255, 0.2)',
                showlegend=False
            ), row=1, col=2)
            
            # Current point
            fig.add_trace(go.Scatter(
                x=[frame], y=[self.loss_curve[frame]],
                mode='markers',
                marker=dict(size=12, color='#ff4444', symbol='circle'),
                showlegend=False
            ), row=1, col=2)
            
            fig.update_layout(
                height=600,
                width=1100,
                showlegend=False,
                paper_bgcolor='rgb(10, 15, 30)',
                plot_bgcolor='rgb(10, 15, 30)',
                margin=dict(l=0, r=10, t=10, b=10),
                scene=dict(
                    xaxis=dict(showgrid=True, gridcolor='rgba(100,150,200,0.1)', showbackground=True,
                              backgroundcolor='rgb(10, 15, 30)', showticklabels=False, title=''),
                    yaxis=dict(showgrid=True, gridcolor='rgba(100,150,200,0.1)', showbackground=True,
                              backgroundcolor='rgb(10, 15, 30)', showticklabels=False, title=''),
                    zaxis=dict(showgrid=True, gridcolor='rgba(100,150,200,0.1)', showbackground=True,
                              backgroundcolor='rgb(10, 15, 30)', showticklabels=False, title=''),
                    camera=dict(eye=dict(x=1.3, y=1.8, z=0.6)),
                    aspectmode='manual',
                    aspectratio=dict(x=2, y=0.5, z=1.2)
                )
            )
            
            fig.update_xaxes(title='Epoch', range=[0, 50], gridcolor='rgba(100,150,200,0.2)',
                            color='#aaa', row=1, col=2)
            fig.update_yaxes(title='Loss', range=[0, 1], gridcolor='rgba(100,150,200,0.2)',
                            color='#aaa', row=1, col=2)
            
            fig.show()
    
    def play_animation(self, _):
        """Play the training animation"""
        import time
        
        if self.is_playing:
            self.is_playing = False
            self.play_btn.description = '▶️ Play'
            return
        
        self.is_playing = True
        self.play_btn.description = '⏸️ Pause'
        
        for i in range(self.epoch_slider.value, len(self.activation_frames)):
            if not self.is_playing:
                break
            self.epoch_slider.value = i
            time.sleep(0.1)
        
        self.is_playing = False
        self.play_btn.description = '▶️ Play'
    
    def reset(self, _):
        """Reset to beginning"""
        self.is_playing = False
        self.play_btn.description = '▶️ Play'
        self.epoch_slider.value = 0

# Create the visualization
NetworkLearningViz()

display(HTML("""
<div class="insight-box">
    <h3 style="margin-top:0; color:#856404;">💡 What You're Seeing</h3>
    <ul style="margin-bottom: 0;">
        <li><strong>Node size & brightness</strong> = activation level (how "active" that neuron is)</li>
        <li><strong>Early training</strong> = chaotic, random activations</li>
        <li><strong>Later training</strong> = organized, meaningful patterns emerge</li>
        <li><strong>Loss decreasing</strong> = network is learning!</li>
    </ul>
    <p style="margin-top: 10px;"><em>Drag the slider or click Play to see the network learn!</em></p>
</div>
"""))
```
